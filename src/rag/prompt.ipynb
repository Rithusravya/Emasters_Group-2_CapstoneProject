```python
%%writefile src/rag/prompt.ipynb

def prompt(task_instruction: str, retrieved_contexts: list, query: str) -> str:
    context_str = "\n".join([f"- Example Context: {c}" for c in retrieved_contexts])
    user_prompt = f"### Task: {task_instruction}\n\n### Relevant Context:\n{context_str}\n\n### Query:\n{query}\n\n### Solution:"
    return user_prompt
```
