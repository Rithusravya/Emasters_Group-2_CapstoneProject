```python
%%writefile src/rag/rag_retriever.ipynb

class RAG:
    def __init__(self, faiss_indexer, ast_indexer):
        self.faiss_indexer = faiss_indexer
        self.ast_indexer = ast_indexer

    def retrieve(self, query_embed, top_k=3):
        return self.faiss_indexer.search(query_embed, top_k=top_k)
```
