```python
from typing import List
from src.models.codegen_loader_small_model import CodeGenModelWrapper
from src.indexing.semantic_index import FAISSIndexManager
```

```python
class TopKRetriever:
    def __init__(self, model: CodeGenModelWrapper, faiss_manager: FAISSIndexManager):
        self.model = model
        self.faiss_manager = faiss_manager

    def retrieve(self, query: str, top_k: int = 2) -> List[str]:
        query_emb = self.model.extract_embeddings(query)
        return self.faiss_manager.search(query_emb, top_k=top_k)
```
