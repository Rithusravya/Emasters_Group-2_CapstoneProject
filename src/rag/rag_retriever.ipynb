```python
from __future__ import annotations
from typing import List
from src.indexing.semantic_index import FAISSIndexManager
from src.models.codegen_loader_small_model import CodeGenModelWrapper
```

```python
class TopKRetriever:
    def __init__(self, model: "CodeGenModelWrapper", faiss_manager: "FAISSIndexManager"):
        self.model = model
        self.faiss_manager = faiss_manager

    def retrieve(self, query: str, top_k: int = 3) -> List[str]:
        query_embedding = self.model.extract_embeddings(query)
        return self.faiss_manager.search(query_embedding, top_k=top_k)
```
