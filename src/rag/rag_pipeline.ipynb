```python
from __future__ import annotations
from typing import List
from src.models.codegen_loader_small_model import CodeGenModelWrapper
```

```python
class RAGPipeline:
    def __init__(self, model: "CodeGenModelWrapper"):
        self.model = model

    def build_prompt(self, query: str, retrieved_context: List[str]) -> str:
        context_str = "\n".join(f"# Retrieved Context/Example:\n{ctx}" for ctx in retrieved_context)
        return (
            f"Below is context retrieved from the repository:\n"
            f"{context_str}\n\n"
            f"# User Request:\n{query}\n"
            f"# Generated Output:\n"
        )

    def execute(self, query: str, retrieved_context: List[str], max_new_tokens: int = 256) -> str:
        prompt = self.build_prompt(query, retrieved_context)
        generated = self.model.generate(prompt, max_new_tokens=max_new_tokens, num_return_sequences=1)
        return generated[0] if generated else ""
```
