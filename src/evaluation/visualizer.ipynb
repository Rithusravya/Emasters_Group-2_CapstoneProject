```python
%%writefile src/evaluation/visualizer.ipynb

import matplotlib.pyplot as plt
import os
```

```python
def plot_evaluation_results(metrics_dict: dict, output_path: str = "outputs/charts/comparison.png"):
    os.makedirs(os.path.dirname(output_path), exist_ok=True)
    models = list(metrics_dict.keys())
    scores = list(metrics_dict.values())

    plt.figure(figsize=(8, 5))
    plt.bar(models, scores, color=['#1f77b4', '#ff7f0e', '#2ca02c'])
    plt.title("Performance Comparison Across Approaches")
    plt.ylabel("Score")
    plt.ylim(0, 1.0)
    plt.grid(axis='y', linestyle='--', alpha=0.7)

    plt.savefig(output_path)
    plt.close()
    print(f"[Step 14] Evaluation chart saved to {output_path}")
```
